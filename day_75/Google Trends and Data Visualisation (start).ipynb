### Converting from Daily to Monthly Data

[Pandas .resample() documentation](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.resample.html) <br>

```python
df_btc_monthly_price = df_btc_price.resample("ME", on="DATE").mean()
print(df_btc_monthly_price.shape)
df_btc_monthly_price.head()
```

```python
df_btc_monthly_price = df_btc_price.resample("ME", on="DATE").last()
print(df_btc_monthly_price.shape)
df_btc_monthly_price.head()
```

# Data Visualisation

### Notebook Formatting & Style Helpers

```python
# Create locators for ticks on the time axis

plt.figure(figsize=(14,8), dpi=200)
plt.xticks(fontsize=10, rotation=45)
plt.xlabel('Month', fontsize=14)
plt.yticks(fontsize=10)
plt.title("Tesla Web Search vs Price", fontsize=20)

# Set the currect axis to use for data visualization
ax1 = plt.gca()
ax1.set_xlim(df_tesla.MONTH.min(),df_tesla.MONTH.max())
ax1.set_ylim(0,600)
# Of the two axis, only x is twin
ax2 = ax1.twinx()

# Set labels to two y label
ax1.set_ylabel("TSLA Stock Price", color="orange", fontsize=14)
ax2.set_ylabel("Search Trend", color="skyblue", fontsize=14)

ax1.plot(df_tesla.MONTH, df_tesla.TSLA_USD_CLOSE, color="orange", linewidth=2)
ax2.plot(df_tesla.MONTH, df_tesla.TSLA_WEB_SEARCH, color="skyblue", linewidth=2)

```
